# Predictive Analytics Using Historical Data

Sales forecasting with Linear Regression.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('../dataset/historical_sales.csv', parse_dates=['Date'])
df.head()
```

```python
df.info()
print('Missing values:\n', df.isna().sum())
df = df.drop_duplicates().dropna(subset=['Date','Sales'])
```

```python
plt.figure(figsize=(10,5))
plt.plot(df['Date'], df['Sales'])
plt.title('Historical Monthly Sales Trend')
plt.xlabel('Date'); plt.ylabel('Sales'); plt.grid(alpha=.25)
plt.show()
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['TimeIndex'] = np.arange(len(df))
features = ['Year','Month','TimeIndex']
X, y = df[features], df['Sales']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, shuffle=False)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
pred = model.predict(X_test)
mae = mean_absolute_error(y_test, pred)
rmse = mean_squared_error(y_test, pred) ** 0.5
r2 = r2_score(y_test, pred)
print('MAE:', round(mae,2))
print('RMSE:', round(rmse,2))
print('R2 Score:', round(r2,4))
```

```python
plt.figure(figsize=(10,5))
plt.plot(X_test.index, y_test.values, label='Actual')
plt.plot(X_test.index, pred, label='Predicted')
plt.title('Actual vs Predicted Sales')
plt.xlabel('Test Observation'); plt.ylabel('Sales'); plt.legend(); plt.grid(alpha=.25)
plt.show()
```

```python
future_dates = pd.date_range(df['Date'].max()+pd.offsets.MonthBegin(1), periods=12, freq='MS')
future = pd.DataFrame({'Date': future_dates})
future['Year'] = future['Date'].dt.year
future['Month'] = future['Date'].dt.month
future['TimeIndex'] = np.arange(len(df), len(df)+12)
future['Predicted_Sales'] = model.predict(future[features])
future
```

```python
plt.figure(figsize=(10,5))
plt.plot(df['Date'], df['Sales'], label='Historical')
plt.plot(future['Date'], future['Predicted_Sales'], '--', label='12-Month Forecast')
plt.title('Future Sales Forecast')
plt.xlabel('Date'); plt.ylabel('Sales'); plt.legend(); plt.grid(alpha=.25)
plt.show()
```
